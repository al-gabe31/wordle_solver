# Importing Modules

```python
from utilities import *
import math
```

# Generating Word List Dictionary

```python
word_list = {}

#* getting list of words
with open('allowed.txt', 'r', encoding='utf-8') as file:
    for line in file:
        word_list[line.strip()] = get_color_dict()

print(len(word_list.keys()))
```

# Getting Color Feedback Frequency for Each Word

```python
words_remaining = len(word_list.keys())

for input_word in word_list.keys():
    for compare_word in word_list.keys():
        curr_feedback = get_color_feedback(input_word, compare_word)
        word_list[input_word][curr_feedback] += 1 #* increment color feedback frequency for input word
    words_remaining -= 1
    print(f'Words Remaining: {words_remaining}')
```

```python
for color_feedback in word_list['slate']:
    print(f'{color_feedback} ==> {word_list["slate"][color_feedback]}')
```

# Calculating Shannon Entropy Scores for Each Word

```python
def get_shannon_entropy(input_dict:dict) -> int:
    total_frequency = 0

    for key in input_dict:
        total_frequency += input_dict[key]

    #* calculating shannon entropy score
    shannon_entropy = 0

    for key in input_dict:
        if input_dict[key] != 0:
            p = input_dict[key]/total_frequency
            shannon_entropy -= p * math.log(p,2)
    
    return shannon_entropy
```

# Getting Top Starter Words

```python
shannon_entropy_scores = []

for word in word_list:
    curr_tuple = (word, get_shannon_entropy(word_list[word]))
    shannon_entropy_scores.append(curr_tuple)

#* sorting the words based on shannon entropy scores from greatest to least
shannon_entropy_scores.sort(key=lambda x: x[1], reverse=True)

print('TOP STARTER WORDS')
for i in range(20):
    print(f'[{i+1}] {shannon_entropy_scores[i][0]} ==> {shannon_entropy_scores[i][1]}')
```
